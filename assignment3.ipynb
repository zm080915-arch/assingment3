```python
import numpy as np
array_1=np.random.randint(1,101,(3,3))
print(array_1)
max=np.max(array_1)
min=np.min(array_1)
print(f"min={min}")
print(f"max={max}")
normalize_array=(array_1-min)/(max-min)
print(f"array after normalize={normalize_array}")
```

```python
import numpy as np
array_2=np.random.randint(1,11,10)
print(array_2)
num_cheeck=int(input("enter the number to cheeck"))
closest_value=array_2[np.abs((array_2-num_cheeck)).argmin()]
print(closest_value)
```
